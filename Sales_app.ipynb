```python
%%writefile sales_app.py
import streamlit as st
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
st.set_page_config(page_title="Sales Predictor",layout="Centered")
st.title('Sales Prediction Tool')
st.write("Enter your advertising budget to estimate expected sales")
np.random.seed(42)
ad_spend=np.linspace(1000,2000,50)
sales=5000+3.5*ad_spend+np.random.normal(0,2000,50)
model=LinearRegression()
model.fit(ad_spend.reshape(-1,1),sales)
budget = st.slider("Select Advertising Budget(rupees)",1000,2000,5000)
prediction = model.predict([[budget]]) [0]
st.subheader(f"Estimated Sales:rupees {prediction:,0f}")
if prediction < 20000:
    suggestion="Sales are relatively low.Consider increasing your advertising budget."
elif prediction < 40000:
    suggestion ="Sales are moderate.Optimizzing your marketing strategy can improve results."
else:
    suggestion="Sales performance is strong.Maintain or strategically scale your budget"
    st.write(suggestion)
```
